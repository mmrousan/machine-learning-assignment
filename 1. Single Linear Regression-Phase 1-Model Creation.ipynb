```python
# This is how we import a library such as Pandas. Pandas contains many built-in functions.
# These functions have already been created by developers and are readily available for use in our daily tasks.
# We shorten the name "pandas" to "pd" using the keyword "as" for convenience.
# Pandas is mainly used to work with tabular data, such as data stored in spreadsheets and databases.
```

```python
import pandas as pd
```

```python
# We are loading the data file saved in the same folder as the Python project.
# The file path must be correct, and the filename must exactly match the name of the data file.
# Here, we create a variable called "dataset" and use the Pandas read_csv() function to read the data.
# When this code is executed, the entire CSV dataset is loaded into the "dataset" variable.
```

```python
dataset=pd.read_csv("Salary_Data.csv")
```

```python
# After creating and assigning a value to a variable, it is good practice to test it to ensure that the assignment was successful.
```

```python
dataset
```

```python
# The next step in model creation/learning is to split the dataset into input and output variables.
# Here, we assign the input column, "YearsExperience", to a variable called "independent".
# Since the original data is already stored in the variable "dataset", we select the "YearsExperience" column from it.
```

```python
independent=dataset[["YearsExperience"]]
```

```python
# It is good practice to test variable to ensure that the assignment was successful.
```

```python
independent
```

```python
# Here, we assign the output column, "Salary", to a variable called "dependent".
# Since the original data is already stored in the variable "dataset", we select the "Salary" column from it.
```

```python
dependent=dataset[["Salary"]]
```

```python
# It is good practice to test the variable to ensure that the assignment was successful.
```

```python
dependent 
```

```python
# # The next step in model creation/learning is "Split train and test".
```

```python
# Similar to Pandas, we use the "sklearn" library for this task.
# However, instead of importing the entire library, we import only the required function.
# Here, we import the train_test_split() function from sklearn's model_selection module.
# This function is used to divide the independent and dependent variables into training and testing datasets.
```

```python
# The train_test_split() function returns four separate datasets as follows:
# X_train – input data used to train the model
# X_test  – input data used to test the model
# y_train – correct output used to train the model
# y_test  – correct output used to test the model
# Therefore, we use four separate variables to store the four results.
# X represents the independent variable, while y represents the dependent variable.
#70% of the data is used for learning, and 30% is used to test what the model has learned.
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test=train_test_split(independent, dependent, test_size=0.30, random_state=0)
```

```python
# We import the LinearRegression model from the sklearn library.
# We create the model and store it in a variable called "regressor".
# The fit() function trains the model using X_train and y_train.
# During training, the model calculates the weight (coefficient) and bias (intercept).
# After this step, the model has learned the relationship between experience and salary.
#The model studies the training data and learns how changes in years of experience affect salary.
```

```python
from sklearn.linear_model import LinearRegression
regressor=LinearRegression()

regressor.fit(X_train,y_train)
```

```python
# The message shown above is a warning, not an error, and the code can still run.
```

```python
# Display the weight (w) learned by the trained model.
# The weight is stored in the regressor's coef_ attribute.
```

```python
weight=regressor.coef_
weight
```

```python
# Display the bias (b) learned by the trained model.
# The bias is stored in the regressor's intercept_ attribute.
```

```python
bais=regressor.intercept_
bais
```

```python
# The trained model uses the testing input data (X_test)
# to predict the corresponding salary values.
# The predicted salaries are stored in the variable "y_pred".

#regressor — the trained linear regression model.
#.predict() — asks the trained model to make predictions.
#X_test — the years-of-experience data kept aside for testing.
#y_pred — the salaries predicted by the model.
```

```python
y_pred=regressor.predict(X_test)
```

```python
#This is the model evaluation stage
# We import the r2_score function to measure the model's performance.
```

```python
# We compare the actual salaries (y_test) with the predicted salaries (y_pred).
# The calculated R² score is stored in the variable "r_score".
```

```python
from sklearn.metrics import r2_score 
r_score=r2_score(y_test,y_pred)
```

```python
# Display the R² score.
#The R² score is close to 1, so the model performs well on the test data. 
#It is like cross-checking whether the model has learned the relationship correctly
```

```python
r_score
```

```python
# We use the pickle library to save the trained model.
# We create a filename called "finalized_model_linear.sav"
# and store it in the variable "filename".
```

```python
#The trained model is saved as a file named finalized_model_linear.sav. 
#We can load and reuse this model later without training it again.
```

```python
import pickle
filename="findlized_model_linear.sav"
```

```python
# We save the trained model stored in "regressor" to the specified file.
# When the following code is executed, the trained model is saved as a file.
# The file will be created in the same folder where the Jupyter Notebook is running.
# It will also appear in Jupyter's file browser because both locations refer to the same folder.
```

```python
pickle.dump(regressor,open(filename,'wb'))
```

```python
# We load the previously saved model from the file.
# "rb" means that the file is opened in read-binary mode.

# We give 13 years of experience to the loaded model.
# The model then predicts the expected salary.
# The predicted result is stored in the variable "result".
```

```python
loaded_model=pickle.load(open("findlized_model_linear.sav",'rb'))
result=loaded_model.predict([[13]])
```
